## Context & Methods

### Key Assumptions
UTC开盘左闭右开；理论完整边界不等于网络到达。此为stdlib顺序回放，不是Jupyter kernel验收。

```python
from pathlib import Path
import json, hashlib
root = Path.cwd()
while not (root/'yoyo').exists() and root != root.parent:
    root = root.parent
here = root/'experiments/active/exp-btcusdtp-genuine-flow-coverage-20260907-v34'
saved = json.loads((here/'report_data.json').read_text())
assert hashlib.sha256((here/'report_data.json').read_bytes()).hexdigest() == 'eb7dc114551400b6fc3470946e273f876a5b1f0c827798be569a3c0334efcbb8'
print('saved SQL result hash: verified')
```

## Data
24个月完整月表；下方仅展示前10个月。源SQL保存在report_data.json的queries。

```python
months = saved['data']['monthly']
assert len(months)==24
print(json.dumps(months[:10], ensure_ascii=False))
```

## Results
核对完整时间分母、方向反例分母和零量条目。

```python
overview = saved['data']['overview'][0]
assert sum(r['valid_bars'] for r in months) == overview['valid_bars'] == 210528
assert sum(r['missing_bars'] for r in months)==0
assert sum(r['opposite_bars'] for r in months)==overview['opposite_bars']==40489
assert overview['nonzero_bars']==209923
assert len(saved['data']['zero_bars'])==38
print(json.dumps(overview, ensure_ascii=False))
```
