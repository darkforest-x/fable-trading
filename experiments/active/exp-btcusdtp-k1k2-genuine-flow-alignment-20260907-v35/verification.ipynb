### Key Assumptions
UTC半开窗口；理论可用不等于收到。无Jupyter依赖，用stdlib顺序执行并记录输出，不冒充kernel验收。

```python
from pathlib import Path
import json, hashlib
root=Path.cwd()
while not (root/'yoyo').exists() and root != root.parent:
    root=root.parent
here=root/'experiments/active/exp-btcusdtp-k1k2-genuine-flow-alignment-20260907-v35'
saved=json.loads((here/'report_data.json').read_text())
assert hashlib.sha256((here/'report_data.json').read_bytes()).hexdigest()=='640030b4c253eda2fa4b48ae98ff0000a8384e14ad298ad1e1d9c15eed4cb511'
print('report data hash verified')
```

## Data
检查全部状态类别和4个半年分段，禁止只挑有效K2或赚钱窗口。

```python
data=saved['data']
assert sum(r['mothers'] for r in data['case_status'])==251
assert sum(r['requests'] for r in data['fold_counts'])==55
print(json.dumps(data['fold_counts'],ensure_ascii=False))
```

## Results
独立核对窗口数与对照母群覆盖；计次不是唯一K线数。

```python
assert sum(r['windows'] for r in data['alignment'])==1001
assert all(r['defined']==r['windows'] for r in data['alignment'])
c=data['control_coverage']
assert sum(r['mothers_with_three_controls'] for r in c)*3==462
print(json.dumps(c,ensure_ascii=False))
```
